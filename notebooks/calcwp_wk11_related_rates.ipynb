## 1. The Idea Behind Related Rates

Suppose two quantities, $x$ and $y$, are related by an equation — say $ x^2 + y^2 = 25 $ (a circle, like Week 7). If both $x$ and $y$ are secretly changing over time (i.e., $x = x(t)$ and $y = y(t)$), then differentiating **both sides with respect to $t$** connects their rates of change:

$$ \frac{d}{dt}[x^2] + \frac{d}{dt}[y^2] = \frac{d}{dt}[25] $$

By the chain rule, $ \dfrac{d}{dt}[x^2] = 2x\dfrac{dx}{dt} $ (since $x$ depends on $t$). So:

$$ 2x\frac{dx}{dt} + 2y\frac{dy}{dt} = 0 $$

This single equation relates $\dfrac{dx}{dt}$ and $\dfrac{dy}{dt}$ — if we know one rate (and the current values of $x$ and $y$), we can solve for the other. This is exactly what "related rates" means: the rates are **related** through the original equation.

Let's set this up in `sympy`, using `Function` for both $x(t)$ and $y(t)$ — the same tool from Week 7's implicit differentiation, just with time as the variable instead of $x$.


````
from sympy import symbols, Function, Eq, diff, solve, Derivative

t = symbols('t')
x = Function('x')(t)
y = Function('y')(t)

circle_eq = Eq(x**2 + y**2, 25)

differentiated = Eq(diff(circle_eq.lhs, t), diff(circle_eq.rhs, t))
print("After differentiating with respect to t:")
print(differentiated)
````

`sympy` produced exactly the equation we derived by hand, with `Derivative(x(t), t)` representing $\dfrac{dx}{dt}$ and `Derivative(y(t), t)` representing $\dfrac{dy}{dt}$.

## 2. Classic Problem: The Sliding Ladder

A 10-meter ladder leans against a wall. The bottom of the ladder is pulled away from the wall at a rate of $2$ m/s. How fast is the top of the ladder sliding **down** the wall when the bottom is $6$ meters from the wall?

**Setup:** Let $x$ = distance from the wall to the base of the ladder, and $y$ = height of the top of the ladder on the wall. Since the ladder has fixed length 10, the Pythagorean theorem gives our connecting equation:

$$ x^2 + y^2 = 10^2 = 100 $$

We're told $\dfrac{dx}{dt} = 2$ (the base moves away at 2 m/s), and we want $\dfrac{dy}{dt}$ at the moment $x=6$.


````
x = Function('x')(t)
y = Function('y')(t)

ladder_eq = Eq(x**2 + y**2, 100)
differentiated = Eq(diff(ladder_eq.lhs, t), diff(ladder_eq.rhs, t))
print("Differentiated equation:", differentiated)

dy_dt = solve(differentiated, Derivative(y, t))[0]
print("dy/dt =", dy_dt)
````

Before we can substitute, we need the value of $y$ when $x=6$. We get this from the **original** equation (not the differentiated one) — at that specific instant, the Pythagorean relationship still holds:

````
from sympy import sqrt

y_value = sqrt(100 - 6**2)
print("y when x=6:", y_value)
````

Now substitute $x=6$, $y=8$, and $\dfrac{dx}{dt}=2$ into our `dy/dt` expression:

````
result = dy_dt.subs({x: 6, y: y_value, Derivative(x, t): 2})
print("dy/dt =", result, "m/s")
````

The negative sign makes sense: as the base slides **away** from the wall, the top slides **down** — a negative rate of change in height. The ladder's top is sliding down at $1.5$ m/s at this instant.

## 3. Classic Problem: The Expanding Circle

A circular oil spill is growing, with its radius increasing at a rate of $3$ m/hour. How fast is the **area** of the spill increasing when the radius is $50$ meters?

**Setup:** The connecting equation is the area formula: $ A = \pi r^2 $, where both $A$ and $r$ depend on time.


````
from sympy import pi

r = Function('r')(t)
A = Function('A')(t)

area_eq = Eq(A, pi * r**2)
differentiated = Eq(diff(area_eq.lhs, t), diff(area_eq.rhs, t))
print("Differentiated equation:", differentiated)
````

Here, `diff(area_eq.lhs, t)` is just $\dfrac{dA}{dt}$ directly (since $A$ is alone on the left), and the right side used the chain rule on $r^2$ automatically. Now substitute $r=50$ and $\dfrac{dr}{dt}=3$:

````
dA_dt_expr = differentiated.rhs
result = dA_dt_expr.subs({r: 50, Derivative(r, t): 3})
print("dA/dt =", result, "square meters per hour")
print("Approximately:", float(result), "m^2/hour")
````

## 4. A Reusable Related Rates Solver

Since every related rates problem follows the same steps (differentiate, then substitute known values), let's write one function that handles this pattern generally — bringing together `Function`, `Eq`, `diff`, and `.subs()` from this week with the `def` habit from Weeks 5, 8, and 9.


````
def solve_related_rate(equation, target_derivative, known_values):
    """
    equation: a sympy Eq relating time-dependent quantities
    target_derivative: the Derivative(...) we want to solve for
    known_values: a dict of substitutions (values and known rates)
    """
    differentiated = Eq(diff(equation.lhs, t), diff(equation.rhs, t))
    solved_expr = solve(differentiated, target_derivative)[0]
    return solved_expr.subs(known_values)

# Re-solve the ladder problem from Section 2 using this reusable function
x = Function('x')(t)
y = Function('y')(t)

ladder_eq = Eq(x**2 + y**2, 100)
known = {x: 6, y: 8, Derivative(x, t): 2}

answer = solve_related_rate(ladder_eq, Derivative(y, t), known)
print("dy/dt =", answer, "m/s")
````
