## 0. Quick Recap from Week 2

Last week, we plotted functions using `numpy` and `matplotlib`, and explored how transformations change a graph's shape. This week, we start using graphs (and tables) for a new purpose: understanding what happens to a function's output as the input gets **very close** to a specific value — without necessarily reaching it. This idea is called a **limit**, and it's the foundation of everything else in Calculus.


## 1. What Is a Limit?

Consider the function:

$$ f(x) = \frac{x^2 - 1}{x - 1} $$

If we try to evaluate $f(1)$ directly, we get $\frac{0}{0}$ — undefined. But that doesn't mean we can't ask: **"What value does $f(x)$ get close to, as $x$ gets close to 1?"**

That question — "what value does the function approach" — is what a **limit** answers. We write it as:

$$ \lim_{x \to 1} f(x) $$

Read as: "the limit of $f(x)$ as $x$ approaches 1."

Notice: this is different from asking "what is $f(1)$?" We're not evaluating *at* 1 — we're asking what happens *near* 1.


````
def f(x):
    return (x**2 - 1) / (x - 1)

# Let's try evaluating exactly at x = 1
f(1)
````

That error confirms $f(1)$ is undefined. But let's see what happens as $x$ gets **close** to 1, without being exactly 1.

````
print(f(0.9))
print(f(0.99))
print(f(0.999))
print(f(1.001))
print(f(1.01))
print(f(1.1))
````

Notice something? As $x$ gets closer and closer to 1 (from either side), $f(x)$ gets closer and closer to **2** — even though $f(1)$ itself doesn't exist. This is exactly what a limit captures:

$$ \lim_{x \to 1} \frac{x^2 - 1}{x - 1} = 2 $$


## 2. Building a Table of Values with a `for` Loop

Typing out `print(f(0.9))`, `print(f(0.99))`, etc. one at a time works, but it's repetitive. A **`for` loop** lets us repeat an action automatically for every item in a list.

```python
for item in my_list:
    # do something with item
```

Let's use a `for` loop to build a proper table of values approaching $x = 1$ from the left side.


````
x_values = [0.9, 0.99, 0.999, 0.9999]

for x in x_values:
    print(f"f({x}) = {f(x)}")
````

We can make the table cleaner and also approach from **both sides** at once:

````
x_values = [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]

print(f"{'x':>10} | {'f(x)':>10}")
print("-" * 23)
for x in x_values:
    print(f"{x:>10} | {f(x):>10.5f}")
````

## 3. Reading Limits from a Graph

Tables are precise, but graphs let us *see* the same behavior at a glance. Let's plot $ f(x) = \dfrac{x^2 - 1}{x - 1} $ and visually confirm the limit we found.

Since $f(x)$ is undefined exactly at $x=1$, we'll mark that point with an open circle (a common convention meaning "the function approaches this point, but doesn't include it").


````
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 4, 400)
x = x[x != 1]     # remove x = 1 to avoid division by zero
y = (x**2 - 1) / (x - 1)

plt.plot(x, y, color='steelblue')
plt.scatter([1], [2], facecolors='none', edgecolors='red', s=100, zorder=5, label="hole at x=1 (undefined)")

plt.title("Graph of f(x) = (x^2 - 1) / (x - 1)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
````

Notice the graph looks like a straight line passing through $y=2$ at $x=1$ — except there's a "hole" there, shown as an open circle. This confirms: even though $f(1)$ is undefined, the **limit** as $x \to 1$ is still 2, because the graph approaches that height from both sides.

## 4. One-Sided Limits

Sometimes a function approaches **different** values depending on which side you approach from. We distinguish:

- **Left-hand limit:** $ \lim_{x \to a^-} f(x) $ — approaching $a$ from values less than $a$
- **Right-hand limit:** $ \lim_{x \to a^+} f(x) $ — approaching $a$ from values greater than $a$

If both one-sided limits agree, the overall (two-sided) limit exists and equals that common value. If they **disagree**, the limit **does not exist** at that point.

Consider the step function:

$$
f(x) =
\begin{cases}
1 & \text{if } x < 2 \\
3 & \text{if } x \geq 2
\end{cases}
$$


````
def step(x):
    if x < 2:
        return 1
    else:
        return 3

left_side  = [1.9, 1.99, 1.999]
right_side = [2.001, 2.01, 2.1]

print("Approaching from the LEFT:")
for x in left_side:
    print(f"  f({x}) = {step(x)}")

print("Approaching from the RIGHT:")
for x in right_side:
    print(f"  f({x}) = {step(x)}")
````

The left-hand limit is **1**, but the right-hand limit is **3** — they disagree. So $ \lim_{x \to 2} f(x) $ **does not exist**, even though the function is perfectly defined at every x-value (including $x=2$ itself, where $f(2)=3$).

Let's visualize this with a graph.

````
x_left = np.linspace(-1, 2, 100, endpoint=False)
x_right = np.linspace(2, 5, 100)

plt.plot(x_left, [1]*len(x_left), color='steelblue')
plt.plot(x_right, [3]*len(x_right), color='steelblue')

plt.scatter([2], [1], facecolors='none', edgecolors='red', s=100, zorder=5)   # open circle (not reached from left)
plt.scatter([2], [3], facecolors='blue', edgecolors='blue', s=60, zorder=5)   # filled circle (actual value)

plt.title("A Function with No Limit at x = 2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
````

## 5. Other Ways a Limit Can Fail to Exist

Besides mismatched one-sided limits, a limit can fail to exist when a function grows without bound (goes to infinity) near a point. Consider:

$$ f(x) = \frac{1}{x} $$

As $x \to 0$, what happens?


````
def reciprocal(x):
    return 1 / x

x_values = [0.1, 0.01, 0.001, -0.001, -0.01, -0.1]

for x in x_values:
    print(f"f({x}) = {reciprocal(x)}")
````

As $x \to 0^+$ (from the right), $f(x) \to +\infty$. As $x \to 0^-$ (from the left), $f(x) \to -\infty$. Since the function doesn't approach a single finite number from either side, $ \lim_{x \to 0} \frac{1}{x} $ **does not exist**.

````
x_pos = np.linspace(0.05, 3, 200)
x_neg = np.linspace(-3, -0.05, 200)

plt.plot(x_pos, 1/x_pos, color='steelblue')
plt.plot(x_neg, 1/x_neg, color='steelblue')

plt.title("Graph of f(x) = 1/x")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.ylim(-20, 20)
plt.grid(True)
plt.show()
````
