## 1. Generating Many x-values with `numpy.linspace`

`numpy` is a Python library for working with arrays of numbers efficiently. The function `linspace(start, stop, num)` creates `num` evenly spaced values between `start` and `stop`.

```python
numpy.linspace(-5, 5, 11)
```
This creates 11 evenly spaced numbers from -5 to 5.


````python
import numpy as np

x_values = np.linspace(-5, 5, 11)
print(x_values)
````

Notice that `x_values` is not a single number — it's an **array** of numbers. The powerful part: math operations apply to the *entire array at once*, no loop needed.

````python
y_values = x_values ** 2   # squares every value in the array at once
print(y_values)
````

## 2. Plotting with `matplotlib`

Now that we can generate x and y arrays, we can plot them with `matplotlib.pyplot`, usually imported as `plt`.

The basic recipe:
```python
plt.plot(x_values, y_values)
plt.show()
```


````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = x ** 2

plt.plot(x, y)
plt.title("Graph of f(x) = x^2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.show()
````

**Why 100 points?** More points make the curve look smooth. Try changing `100` to `5` in the cell above and re-run it — notice how jagged the graph becomes. This is the same idea we'll revisit later when approximating curves numerically.

## 3. Transformation #1 — Vertical & Horizontal Shifts

Given a base function $ f(x) $, we can shift its graph:

| Transformation | Effect |
|---|---|
| $ f(x) + k $ | Shifts graph **up** by $k$ (if $k>0$) or down (if $k<0$) |
| $ f(x - h) $ | Shifts graph **right** by $h$ (if $h>0$) or left (if $h<0$) |

Let's compare $ f(x) = x^2 $ with $ f(x) + 3 $ and $ f(x - 2) $ on the same plot.


````python
x = np.linspace(-6, 6, 200)

y1 = x ** 2            # original
y2 = x ** 2 + 3         # shifted up by 3
y3 = (x - 2) ** 2       # shifted right by 2

plt.plot(x, y1, label="f(x) = x^2")
plt.plot(x, y2, label="f(x) + 3")
plt.plot(x, y3, label="f(x - 2)")

plt.title("Vertical and Horizontal Shifts")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
````

Notice:
- The **orange** curve (`f(x) + 3`) is the same shape as the blue curve, just moved **up**.
- The **green** curve (`f(x - 2)`) is the same shape, moved **right**.

## 4. Transformation #2 — Vertical Stretch/Compression

Multiplying a function by a constant $a$ changes how "tall" or "flat" it looks:

$$ y = a \cdot f(x) $$

- If $ |a| > 1 $: the graph stretches **vertically** (looks taller/narrower)
- If $ 0 < |a| < 1 $: the graph compresses **vertically** (looks flatter/wider)
- If $ a $ is negative: the graph also **reflects** (flips) — covered next


````python
x = np.linspace(-4, 4, 200)

y1 = x ** 2          # original
y2 = 3 * x ** 2       # stretched (steeper)
y3 = 0.3 * x ** 2     # compressed (flatter)

plt.plot(x, y1, label="f(x) = x^2")
plt.plot(x, y2, label="3f(x)")
plt.plot(x, y3, label="0.3f(x)")

plt.title("Vertical Stretch and Compression")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
````

## 5. Transformation #3 — Reflections

- $ -f(x) $ reflects the graph **over the x-axis** (flips upside down)
- $ f(-x) $ reflects the graph **over the y-axis** (flips left-right)


````python
x = np.linspace(-4, 4, 200)

y1 = x ** 3          # original
y2 = -(x ** 3)        # reflected over x-axis
y3 = (-x) ** 3        # reflected over y-axis

plt.plot(x, y1, label="f(x) = x^3")
plt.plot(x, y2, label="-f(x)")
plt.plot(x, y3, label="f(-x)")

plt.title("Reflections")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
````

**Notice something interesting?** For $ f(x) = x^3 $, the graphs of $ -f(x) $ and $ f(-x) $ actually look identical! This is a special property of **odd functions** — you don't need to memorize this term yet, just observe it. Not every function behaves this way (try it with $x^2$ instead and see what happens).

## 6. Combining Transformations

Real transformations often combine several of these at once:

$$ y = a \cdot f(x - h) + k $$

- $a$: vertical stretch/compression (and reflection if negative)
- $h$: horizontal shift
- $k$: vertical shift

**Worked Example:** Starting from $ f(x) = x^2 $, graph $ y = -2(x - 1)^2 + 4 $.

Reading the transformation:
- $a = -2$: stretched by 2, and reflected (flipped upside down)
- $h = 1$: shifted right by 1
- $k = 4$: shifted up by 4


````python
x = np.linspace(-4, 6, 200)

y_original = x ** 2
y_transformed = -2 * (x - 1) ** 2 + 4

plt.plot(x, y_original, label="f(x) = x^2", linestyle='--')
plt.plot(x, y_transformed, label="-2(x-1)^2 + 4")

plt.title("Combined Transformation")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
````
