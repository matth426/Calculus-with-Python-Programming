## 1. Average Rate of Change

Suppose a car's position (in meters) at time $t$ (in seconds) is given by $ s(t) = t^2 $.

Between $t=1$ and $t=3$, how fast did the car travel, *on average*?

$$ \text{average rate of change} = \frac{s(3) - s(1)}{3 - 1} = \frac{9 - 1}{2} = 4 \text{ m/s} $$

In general, the **average rate of change** of $f(x)$ over an interval $[a, b]$ is:

$$ \frac{f(b) - f(a)}{b - a} $$

Graphically, this is the **slope of the secant line** — the straight line connecting the two points $(a, f(a))$ and $(b, f(b))$ on the curve.


````
def s(t):
    return t ** 2

avg_rate = (s(3) - s(1)) / (3 - 1)
print("Average rate of change from t=1 to t=3:", avg_rate)
````

Notice we just wrote our first **reusable function** using `def`:

```python
def s(t):
    return t ** 2
```

This defines a function named `s` that takes one input (`t`) and returns `t ** 2`. Once defined, we can call `s(1)`, `s(3)`, `s(100)` — anything — without rewriting the formula each time. This is much more convenient than the "assign x, then recompute" approach from Weeks 1–4.

## 2. Visualizing the Secant Line

Let's plot $ s(t) = t^2 $ along with the secant line through $t=1$ and $t=3$, to see what "average rate of change" looks like geometrically.


````
import numpy as np
import matplotlib.pyplot as plt

def s(t):
    return t ** 2

t = np.linspace(0, 4, 200)
y = s(t)

# secant line through (1, s(1)) and (3, s(3))
slope = (s(3) - s(1)) / (3 - 1)
secant_line = s(1) + slope * (t - 1)

plt.plot(t, y, label="s(t) = t^2")
plt.plot(t, secant_line, '--', color='orange', label=f"secant line (slope = {slope})")
plt.scatter([1, 3], [s(1), s(3)], color='red', zorder=5)

plt.title("Average Rate of Change as the Secant Line's Slope")
plt.xlabel("t")
plt.ylabel("s(t)")
plt.legend()
plt.grid(True)
plt.show()
````

## 3. From Average to Instantaneous — The Difference Quotient

The average rate of change used two *distant* points. But what if we want the rate of change **at a single instant** — say, exactly at $t=1$, not averaged between $t=1$ and $t=3$?

The idea: instead of a fixed second point like $t=3$, use a point that's just *a tiny bit away* from $t=1$ — call the tiny distance $h$. This gives the **difference quotient**:

$$ \frac{f(a+h) - f(a)}{h} $$

This is still an average rate of change — just over a much smaller interval, from $a$ to $a+h$. As $h$ gets smaller and smaller (approaching 0), this ratio approaches the **instantaneous rate of change** at $x=a$ — which is exactly the **derivative**, written $f'(a)$.

$$ f'(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h} $$

We can't literally set $h=0$ (that would divide by zero), but we can get a very good *numerical approximation* by choosing a very small $h$, like $0.001$.


````
def difference_quotient(f, a, h):
    return (f(a + h) - f(a)) / h

def s(t):
    return t ** 2

# Approximate the derivative of s(t) = t^2 at t = 1
approx = difference_quotient(s, a=1, h=0.001)
print("Approximate derivative at t=1:", approx)
````

We wrote another `def` function — this time, `difference_quotient` takes **three** inputs: a function `f`, a point `a`, and a small step `h`. Notice that `f` itself is passed in as an argument — in Python, functions can be passed around just like numbers or strings.

Let's see what happens as we make `h` smaller and smaller — does the approximation stabilize?


````
for h in [1, 0.1, 0.01, 0.001, 0.0001, 0.00001]:
    approx = difference_quotient(s, a=1, h=h)
    print(f"h = {h:<10} approximate derivative = {approx}")
````

As $h \to 0$, the approximation settles at **2**. This matches what you may already suspect: for $s(t) = t^2$, the derivative formula is $s'(t) = 2t$, so $s'(1) = 2$. We'll prove this properly (and compute it symbolically) next week — for now, we're building the numerical intuition.

## 4. From Secant to Tangent

As $h$ shrinks, the secant line (connecting two points) starts to look like the **tangent line** — the line that just barely touches the curve at a single point, matching its exact slope there.

Let's animate this idea by plotting secant lines for shrinking values of $h$, all at once.


````
t = np.linspace(0, 3, 200)
y = s(t)

plt.plot(t, y, label="s(t) = t^2", color='steelblue', linewidth=2)

a = 1
for h in [1.5, 0.8, 0.3]:
    slope = difference_quotient(s, a, h)
    secant = s(a) + slope * (t - a)
    plt.plot(t, secant, '--', label=f"secant, h={h} (slope={slope:.2f})")

# The tangent line (using a very small h as a stand-in for the true derivative)
true_slope = difference_quotient(s, a, 0.00001)
tangent = s(a) + true_slope * (t - a)
plt.plot(t, tangent, color='red', linewidth=2, label=f"tangent line (slope≈{true_slope:.2f})")

plt.scatter([a], [s(a)], color='black', zorder=5)
plt.ylim(-1, 8)
plt.title("Secant Lines Approaching the Tangent Line as h Shrinks")
plt.xlabel("t")
plt.ylabel("s(t)")
plt.legend()
plt.grid(True)
plt.show()
````

Notice how the secant lines rotate and get closer to the red tangent line as $h$ shrinks. This is the geometric picture behind the derivative: **the derivative at a point is the slope of the tangent line there.**
