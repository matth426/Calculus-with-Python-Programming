## Weeks 5–7: The Derivative, Differentiation Rules, Higher-Order & Implicit

| Concept | Key tool |
|---|---|
| Numerical derivative | Difference quotient: `(f(a+h) - f(a)) / h` for small `h` |
| Exact derivative | `sympy.diff(expr, x)` |
| Power/sum/product/quotient/chain rules | All handled automatically by `diff()` |
| Higher-order derivatives | `diff(expr, x, n)` or nested `diff(diff(expr, x), x)` |
| Implicit differentiation | `Function('y')(x)`, `Eq(...)`, differentiate both sides, `solve(..., Derivative(y, x))` |


````
f = x**3 * (x + 2)**2
print("f'(x) =", diff(f, x))
print("f''(x) =", diff(f, x, 2))
````

**Quick Check 2:** Find $\dfrac{dy}{dx}$ for the implicit equation $ y^3 + xy = 10 $.

````
# Write your code here
````

<details>
<summary><b>Click for Solution</b></summary>

```python
y = Function('y')(x)
eq = Eq(y**3 + x*y, 10)
differentiated = Eq(diff(eq.lhs, x), diff(eq.rhs, x))
dy_dx = solve(differentiated, Derivative(y, x))[0]
print("dy/dx =", dy_dx)
```

</details>

## Week 8: Motion Problems

| Concept | Key tool |
|---|---|
| Position → velocity → acceleration | $v(t) = s'(t)$, $a(t) = v'(t) = s''(t)$ |
| Direction of motion | Sign of $v(t)$: positive = forward, negative = backward |
| Speeding up / slowing down | Same sign (v, a) = speeding up; opposite signs = slowing down |
| Distance vs. displacement | Displacement = $s(b)-s(a)$; total distance sums absolute legs between rest points |


**Quick Check 3:** For $ s(t) = t^3 - 6t^2 $, find $v(t)$ and determine whether the object is speeding up or slowing down at $t=1$.

````
# Write your code here
````

<details>
<summary><b>Click for Solution</b></summary>

```python
t = symbols('t')
s = t**3 - 6*t**2
v = diff(s, t)
a = diff(s, t, 2)

v_at_1 = v.subs(t, 1)
a_at_1 = a.subs(t, 1)
print("v(1) =", v_at_1, ", a(1) =", a_at_1)
```

**Answer:** $v(1)=-9$ (negative), $a(1)=-6$ (negative) — same sign, so the object is **speeding up**.

</details>

## Week 9: Curve Sketching

| Concept | Key tool |
|---|---|
| Critical points | `solve(diff(f, x), x)` |
| Increasing/decreasing | Sign of $f'(x)$ at test points between critical points |
| First Derivative Test | Compare sign of $f'$ just before/after a critical point |
| Inflection points | `solve(diff(f, x, 2), x)`, check for a sign change in $f''$ |


**Quick Check 4:** Find the critical points of $ f(x) = x^3 - 6x^2 + 9x $ and classify each as a local max or local min.

````
# Write your code here
````

<details>
<summary><b>Click for Solution</b></summary>

```python
x = symbols('x')
f = x**3 - 6*x**2 + 9*x
f_prime = diff(f, x)
critical_pts = solve(f_prime, x)
print("Critical points:", critical_pts)

for c in critical_pts:
    before = f_prime.subs(x, c - 0.1)
    after = f_prime.subs(x, c + 0.1)
    kind = "LOCAL MAX" if before > 0 and after < 0 else "LOCAL MIN"
    print(f"x={c}: {kind}")
```

**Answer:** $x=1$ is a local maximum, $x=3$ is a local minimum.

</details>

## Week 10: Optimization

| Concept | Key tool |
|---|---|
| Absolute max/min on $[a,b]$ | Evaluate $f$ at critical points AND endpoints; compare |
| Word problems | Constraint equation → reduce to one variable → optimize |
| Numerical optimization | `scipy.optimize.minimize_scalar(f, bounds=(a,b), method='bounded')` |
| Finding a maximum with `minimize_scalar` | Minimize $-f(x)$ instead, then negate the result |


**Quick Check 5:** Find the absolute maximum of $ f(x) = -x^2 + 4x + 1 $ on $[0, 5]$.

````
# Write your code here
````

<details>
<summary><b>Click for Solution</b></summary>

```python
x = symbols('x')
f = -x**2 + 4*x + 1
f_prime = diff(f, x)
critical_pts = solve(f_prime, x)

candidates = [c for c in critical_pts if 0 <= c <= 5] + [0, 5]
for c in candidates:
    print(f"f({c}) = {f.subs(x, c)}")
```

**Answer:** Absolute maximum is $f(2)=5$.

</details>

## Week 11: Related Rates

| Concept | Key tool |
|---|---|
| Setting up | `Function('x')(t)` for each time-dependent quantity |
| Differentiating | `diff(equation.lhs, t)` and `diff(equation.rhs, t)`, wrapped in `Eq(...)` |
| Solving for a rate | `solve(differentiated_eq, Derivative(target, t))[0]` |
| Missing values | Always pull from the **original** (undifferentiated) equation |


**Quick Check 6:** A square's side length $s$ is increasing at $2$ cm/s. How fast is the area $A=s^2$ increasing when $s=5$ cm?

````
# Write your code here
````

<details>
<summary><b>Click for Solution</b></summary>

```python
t = symbols('t')
s = Function('s')(t)
A = Function('A')(t)

area_eq = Eq(A, s**2)
differentiated = Eq(diff(area_eq.lhs, t), diff(area_eq.rhs, t))
dA_dt = differentiated.rhs.subs({s: 5, Derivative(s, t): 2})
print("dA/dt =", dA_dt, "cm^2/s")
```

**Answer:** 20 cm²/s

</details>
