## 3. Functions — The Building Block of Calculus

A **function** is a rule that assigns exactly one output to each input.

We write:
$$ f(x) = x^2 + 3 $$

This means: "take a number $x$, square it, then add 3."

To **evaluate** a function at a specific value, we substitute that value for $x$. For example:

$$ f(2) = (2)^2 + 3 = 4 + 3 = 7 $$

### Domain and Range
- The **domain** is the set of all valid inputs ($x$-values) for the function.
- The **range** is the set of all possible outputs ($y$-values) the function can produce.

For $f(x) = x^2 + 3$:
- Domain: all real numbers (any $x$ works)
- Range: $y \geq 3$ (since $x^2 \geq 0$, adding 3 means the smallest output is 3)


### Translating a Math Function into Python

We can write $f(x) = x^2 + 3$ as a Python expression and evaluate it manually first, before we learn to write reusable functions with `def` (coming in Week 5).

For now, we'll evaluate it "by hand" using variables:


````python
x = 2
f_x = x ** 2 + 3
print("f(2) =", f_x)
````

````python
# Let's evaluate f(x) = x^2 + 3 at several values of x

x = -3
print("f(-3) =", x ** 2 + 3)

x = 0
print("f(0) =", x ** 2 + 3)

x = 4
print("f(4) =", x ** 2 + 3)
````

## 4. A Sneak Peek at Graphing

Next week, we'll properly learn how to plot functions with the `matplotlib` library. As a preview, here's what a simple graph looks like — you don't need to understand every line yet, just run it and observe.


````python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.linspace(-5, 5, 100)   # 100 evenly spaced x-values from -5 to 5
y_values = x_values ** 2 + 3          # apply f(x) = x^2 + 3 to all of them

plt.plot(x_values, y_values)
plt.title("Graph of f(x) = x^2 + 3")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.show()
````

Notice how the graph never dips below $y = 3$ — this matches the range we found earlier ($y \geq 3$)!
