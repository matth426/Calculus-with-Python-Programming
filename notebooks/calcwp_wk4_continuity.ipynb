## 0. Quick Recap from Week 3

Last week, we estimated limits by building tables of values and reading graphs — a numerical, "get close and see what happens" approach. That works, but it's slow and only gives an *estimate*. This week, we learn **limit laws**, which let us compute limits algebraically (exactly, not just approximately) — and we'll let `sympy` do the algebra for us.


## 1. Limit Laws

If $ \lim_{x \to a} f(x) $ and $ \lim_{x \to a} g(x) $ both exist, then:

</br>

**Constant multiple:**
$$ \lim_{x \to a} [c \cdot f(x)] = c \cdot \lim_{x \to a} f(x) $$

**Sum:**
$$ \lim_{x \to a} [f(x) + g(x)] = \lim_{x \to a} f(x) + \lim_{x \to a} g(x) $$

**Difference:**
$$ \lim_{x \to a} [f(x) - g(x)] = \lim_{x \to a} f(x) - \lim_{x \to a} g(x) $$

**Product:**
$$ \lim_{x \to a} [f(x) \cdot g(x)] = \lim_{x \to a} f(x) \cdot \lim_{x \to a} g(x) $$

**Quotient** (if denominator limit $\neq 0$):
$$ \lim_{x \to a} \frac{f(x)}{g(x)} = \frac{\lim_{x \to a} f(x)}{\lim_{x \to a} g(x)} $$

</br>

These laws mean: for most "nice" functions (polynomials, rational functions where the denominator isn't zero), you can find the limit by **direct substitution** — just plug in $a$.

</br>

**Example:** $ \lim_{x \to 3} (2x^2 - 5x + 1) $

</br>

Since this is a polynomial, we can substitute directly:
$$ 2(3)^2 - 5(3) + 1 = 18 - 15 + 1 = 4 $$

## 2. When Direct Substitution Doesn't Work

Recall from Week 3: $ \lim_{x \to 1} \dfrac{x^2-1}{x-1} $. If we substitute $x=1$ directly, we get $\frac{0}{0}$ — undefined. Direct substitution fails here because the denominator is zero at that point.

In cases like this, we can often **simplify algebraically first** (factor and cancel), then substitute:

$$ \frac{x^2-1}{x-1} = \frac{(x-1)(x+1)}{x-1} = x+1 \quad \text{(for } x \neq 1\text{)} $$

Now substitute directly into the simplified form: $ 1 + 1 = 2 $ — matching what we found by tables last week!

This is where **`sympy`** becomes powerful: it can factor, simplify, and compute limits symbolically, without us doing the algebra by hand.


## 3. Introducing `sympy`

`sympy` is a Python library for **symbolic math** — it works with $x$ as an actual mathematical symbol, not just a number. This lets it do algebra, calculus, and simplification exactly (no rounding, no approximation).

First, we need to declare our variable as a **symbol**:
```python
from sympy import symbols
x = symbols('x')
```

Then we can define expressions using that symbol, and use `sympy.limit()` to compute a limit directly:
```python
from sympy import limit
limit(expression, x, a)   # limit of expression as x -> a
```


````
from sympy import symbols, limit

x = symbols('x')
expression = (x**2 - 1) / (x - 1)

result = limit(expression, x, 1)
print(result)
````

No tables, no manual factoring — `sympy` computed the exact limit directly, and it matches our earlier answer of 2.

Let's also confirm Week 3's other example, $ \lim_{x \to 3} \dfrac{x^2-9}{x-3} $:

````
expression2 = (x**2 - 9) / (x - 3)
result2 = limit(expression2, x, 3)
print(result2)
````

## 4. Continuity

A function $f(x)$ is **continuous** at $x = a$ if **all three** of these conditions hold:

1. $f(a)$ is defined (the function has an actual value at $a$)
2. $ \lim_{x \to a} f(x) $ exists (the function approaches a single value from both sides)
3. $ \lim_{x \to a} f(x) = f(a) $ (the limit matches the actual function value)

If **any** of these fail, the function is **discontinuous** at $a$.

Recall $ f(x) = \dfrac{x^2-1}{x-1} $ from before: condition 1 fails (it's undefined at $x=1$), so it's discontinuous there — even though the limit exists and equals 2.


### Checking Continuity in Python with `if/else`

Let's write a Python check that tests all three conditions at once, using `sympy` to compute the pieces.


````
from sympy import symbols, limit, oo, nan

def check_continuity(expr, x, a):
    # Condition 1: is f(a) defined?
    try:
        f_a = expr.subs(x, a)
        if f_a == nan or f_a.has(oo):
            defined = False
        else:
            defined = True
    except Exception:
        defined = False

    # Condition 2: does the limit exist?
    lim_value = limit(expr, x, a)
    limit_exists = not lim_value.has(oo) and lim_value != nan

    # Condition 3: does the limit equal f(a)?
    matches = defined and limit_exists and (lim_value == f_a)

    print(f"f({a}) defined?      {defined}")
    print(f"Limit exists?        {limit_exists} (value: {lim_value})")
    print(f"Limit equals f({a})? {matches}")

    if defined and limit_exists and matches:
        print(f"\nCONCLUSION: f is CONTINUOUS at x = {a}")
    else:
        print(f"\nCONCLUSION: f is DISCONTINUOUS at x = {a}")
````

````
x = symbols('x')
expr = (x**2 - 1) / (x - 1)

check_continuity(expr, x, 1)
````

As expected: discontinuous at $x=1$, because $f(1)$ is undefined (condition 1 fails), even though the limit exists.

Let's test a function that *is* continuous everywhere, like a simple polynomial:

````
expr2 = x**2 + 3

check_continuity(expr2, x, 1)
````
