## 0. Quick Recap from Week 9

Last week, we used critical points and the First Derivative Test to classify **local** maxima and minima — points that are highest or lowest *compared to their immediate neighbors*. This week, we ask a more practical question: over a specific range of values, what is the **single best** (largest or smallest) outcome? This is **optimization** — arguably the most-used application of Calculus in the real world, from minimizing cost to maximizing area to minimizing material.


## 1. Absolute Maximum and Minimum on a Closed Interval

A **local** max/min (Week 9) is only the best point *nearby*. An **absolute** max/min is the best point across an entire interval — and it might occur at a critical point, OR at one of the interval's endpoints.

**The Closed Interval Method:**
1. Find all critical points of $f(x)$ inside the interval $[a, b]$.
2. Evaluate $f$ at each critical point AND at both endpoints $a$ and $b$.
3. The largest value found is the absolute maximum; the smallest is the absolute minimum.

**Example:** Find the absolute max/min of $ f(x) = x^3 - 6x^2 + 9x + 2 $ on $[0, 5]$.


````
from sympy import symbols, diff, solve

x = symbols('x')
f = x**3 - 6*x**2 + 9*x + 2

f_prime = diff(f, x)
critical_points = solve(f_prime, x)
print("Critical points:", critical_points)

# Keep only critical points that fall INSIDE the interval [0, 5]
interval = (0, 5)
critical_points_in_range = [c for c in critical_points if interval[0] <= c <= interval[1]]
print("Critical points inside [0, 5]:", critical_points_in_range)
````

Now we evaluate $f$ at every candidate: the critical points inside the interval, and both endpoints.

````
candidates = critical_points_in_range + list(interval)

print(f"{'x':>6} | {'f(x)':>8}")
print("-" * 17)
for c in candidates:
    print(f"{str(c):>6} | {str(f.subs(x, c)):>8}")
````

Reading the table: the largest value is the **absolute maximum**, the smallest is the **absolute minimum**, over this specific interval.

## 2. Setting Up a Word Problem

The hardest part of optimization is usually **translating words into a function** — the calculus itself is often the easy part. Here's the general recipe:

1. Identify what quantity needs to be maximized or minimized (this becomes your function).
2. Identify any constraint connecting the variables (this lets you reduce to a single variable).
3. Substitute the constraint into your function so it depends on only one variable.
4. Use the Closed Interval Method (or just find critical points, if the domain isn't a closed interval) to optimize.

**Example:** A farmer has 200 meters of fencing and wants to enclose a rectangular field, using a wall as one side (so fencing is only needed on 3 sides). What dimensions maximize the enclosed area?

Let $x$ = the width of the two sides perpendicular to the wall, and $y$ = the length of the side parallel to the wall.

- **Constraint** (total fencing used): $ 2x + y = 200 $, so $ y = 200 - 2x $
- **Quantity to maximize** (area): $ A = x \cdot y = x(200 - 2x) = 200x - 2x^2 $


````
x = symbols('x')
A = x * (200 - 2*x)
A_expanded = A.expand()
print("Area as a function of x:", A_expanded)
````

Now we optimize, just like Section 1 — except here the domain is naturally bounded by common sense: $x$ must be between 0 and 100 (beyond $x=100$, $y$ would become negative, which isn't physically possible).

````
A_prime = diff(A_expanded, x)
critical_pts = solve(A_prime, x)
print("A'(x) =", A_prime)
print("Critical points:", critical_pts)

# Check endpoints and critical point(s) within [0, 100]
interval = (0, 100)
candidates = [c for c in critical_pts if interval[0] <= c <= interval[1]] + list(interval)

print(f"\n{'x':>6} | {'Area':>8}")
print("-" * 19)
for c in candidates:
    print(f"{str(c):>6} | {str(A_expanded.subs(x, c)):>8}")
````

The maximum area occurs at $x=50$. The corresponding $y$ value and maximum area:

````
best_x = 50
best_y = 200 - 2*best_x
max_area = A_expanded.subs(x, best_x)

print(f"Optimal width (x): {best_x} m")
print(f"Optimal length (y): {best_y} m")
print(f"Maximum area: {max_area} square meters")
````

## 3. Introducing `scipy.optimize`

So far, we've solved optimization problems **symbolically** — finding an exact answer using algebra. But many real-world functions are too complicated to solve by hand (or don't have a clean symbolic solution at all). For these cases, we can use `scipy.optimize`, which finds optimal values **numerically** — searching for the answer rather than deriving it algebraically.

The function `minimize_scalar()` finds where a function reaches its **minimum**. To find a **maximum** instead, we minimize the *negative* of the function (since the highest point of $f(x)$ is the lowest point of $-f(x)$).

Let's verify our farmer's fence problem (Section 2) using `scipy` instead of `sympy`.


````
from scipy.optimize import minimize_scalar

def area(x):
    return -(x * (200 - 2*x))   # negative, because we're minimizing to find a MAXIMUM

result = minimize_scalar(area, bounds=(0, 100), method='bounded')

print("Optimal x (scipy):", result.x)
print("Maximum area (scipy):", -result.fun)   # negate back to get the actual (positive) area
````

Notice `scipy` gives $x \approx 50$, matching our exact symbolic answer from Section 2 — but as a decimal approximation rather than an exact value. `bounds=(0, 100)` tells `scipy` to only search within that interval, and `method='bounded'` is the standard choice when you have a known closed interval to search within.

## 4. Visualizing the Optimum

Let's plot the area function from Section 2 and mark the maximum we found, to see the optimization visually.


````
import numpy as np
import matplotlib.pyplot as plt
from sympy import lambdify

x = symbols('x')
A_expr = x * (200 - 2*x)
A_func = lambdify(x, A_expr, 'numpy')

x_vals = np.linspace(0, 100, 200)
y_vals = A_func(x_vals)

plt.plot(x_vals, y_vals, color='steelblue', label="A(x) = x(200 - 2x)")
plt.scatter([50], [A_func(50)], color='red', zorder=5)
plt.annotate(f"Maximum\n(50, {A_func(50)})", xy=(50, A_func(50)), xytext=(55, 4200),
             arrowprops=dict(arrowstyle='->', color='red'), color='red')

plt.title("Area vs. Width (Fencing Problem)")
plt.xlabel("x (width in meters)")
plt.ylabel("Area (square meters)")
plt.legend()
plt.grid(True)
plt.show()
````

The graph confirms what both `sympy` and `scipy` found algebraically/numerically: the parabola-shaped area function peaks exactly at $x=50$.
