## 0. Quick Recap from Week 8

Over the past month, we've built a full toolkit: derivatives (Weeks 5–7), and using them to analyze motion (Week 8) — direction, rest points, speeding up/slowing down. This week, we recognize that **the exact same logic applies to any function**, not just position over time. This is **curve sketching** — using derivatives to understand a graph's shape without needing to plot hundreds of points.


## 1. Increasing, Decreasing, and Critical Points

Just like velocity told us the direction of motion, $f'(x)$ tells us the direction a curve is heading:

- $ f'(x) > 0 $: the function is **increasing**
- $ f'(x) < 0 $: the function is **decreasing**
- $ f'(x) = 0 $ (or undefined): a **critical point** — a candidate for a local max, local min, or neither

**Example:** $ f(x) = x^3 - 3x^2 - 9x + 5 $


````
from sympy import symbols, diff, solve

x = symbols('x')
f = x**3 - 3*x**2 - 9*x + 5

f_prime = diff(f, x)
critical_points = solve(f_prime, x)

print("f'(x) =", f_prime)
print("Critical points at x =", critical_points)
````

To find increasing/decreasing intervals, we test the sign of $f'(x)$ in the regions **between** critical points (the same "test point" strategy from Week 8's direction-of-motion analysis).

````
test_points = [-3, 0, 5]   # one point before -1, one between -1 and 3, one after 3

for tp in test_points:
    slope = f_prime.subs(x, tp)
    trend = "increasing" if slope > 0 else ("decreasing" if slope < 0 else "flat")
    print(f"At x={tp}: f'(x) = {slope}  -->  {trend}")
````

## 2. The First Derivative Test

Once we know where $f'(x)$ changes sign, we can classify each critical point:

- If $f'(x)$ goes from **positive to negative** at $x=c$: $f$ has a **local maximum** at $x=c$
- If $f'(x)$ goes from **negative to positive** at $x=c$: $f$ has a **local minimum** at $x=c$
- If $f'(x)$ doesn't change sign: $x=c$ is **neither** (e.g., a flat inflection point)

Let's automate this by writing a function that checks the sign of $f'(x)$ just before and just after each critical point.


````
def classify_critical_points(f_prime, critical_pts, var, delta=0.1):
    for c in critical_pts:
        before = f_prime.subs(var, c - delta)
        after = f_prime.subs(var, c + delta)

        if before > 0 and after < 0:
            kind = "LOCAL MAXIMUM"
        elif before < 0 and after > 0:
            kind = "LOCAL MINIMUM"
        else:
            kind = "NEITHER (no sign change)"

        print(f"x = {c}: f' goes from {before:.3f} to {after:.3f}  -->  {kind}")

f = x**3 - 3*x**2 - 9*x + 5
f_prime = diff(f, x)
critical_points = solve(f_prime, x)

classify_critical_points(f_prime, critical_points, x)
````

Notice `classify_critical_points()` takes `delta` as an optional parameter with a **default value** (`delta=0.1`) — if we don't specify it when calling the function, Python automatically uses 0.1. We could call `classify_critical_points(f_prime, critical_points, x, delta=0.01)` to use a smaller test distance instead.

## 3. Concavity and Inflection Points

The **second derivative** tells us how the curve bends:

- $ f''(x) > 0 $: the curve is **concave up** (shaped like a cup, ∪)
- $ f''(x) < 0 $: the curve is **concave down** (shaped like a cap, ∩)
- $ f''(x) = 0 $ (and changes sign): an **inflection point** — where concavity switches

This is found exactly the same way as before, but one derivative order higher: solve $f''(x)=0$, then test points on either side.


````
f_double_prime = diff(f, x, 2)
inflection_candidates = solve(f_double_prime, x)

print("f''(x) =", f_double_prime)
print("Inflection point candidates at x =", inflection_candidates)

for tp in [0, 2]:
    concavity_value = f_double_prime.subs(x, tp)
    shape = "concave up" if concavity_value > 0 else ("concave down" if concavity_value < 0 else "flat")
    print(f"At x={tp}: f''(x) = {concavity_value}  -->  {shape}")
````

## 4. Putting It All Together: An Annotated Curve Sketch

Now let's combine everything — critical points, their classification, and inflection points — into a single annotated graph using `matplotlib`. We'll use `plt.annotate()` to label each feature directly on the curve.


````
import numpy as np
import matplotlib.pyplot as plt
from sympy import lambdify

f = x**3 - 3*x**2 - 9*x + 5
f_prime = diff(f, x)
f_double_prime = diff(f, x, 2)

critical_points = solve(f_prime, x)
inflection_points = solve(f_double_prime, x)

f_func = lambdify(x, f, 'numpy')

x_vals = np.linspace(-4, 6, 400)
y_vals = f_func(x_vals)

plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, color='steelblue', label="f(x) = x^3 - 3x^2 - 9x + 5")
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)

# Annotate critical points
for c in critical_points:
    y_c = f.subs(x, c)
    before = f_prime.subs(x, c - 0.1)
    after = f_prime.subs(x, c + 0.1)
    label = "Local Max" if (before > 0 and after < 0) else "Local Min"

    plt.scatter([c], [y_c], color='red', zorder=5)
    plt.annotate(label, xy=(c, y_c), xytext=(c + 0.5, y_c + 8),
                 arrowprops=dict(arrowstyle='->', color='red'), color='red')

# Annotate inflection points
for ip in inflection_points:
    y_ip = f.subs(x, ip)
    plt.scatter([ip], [y_ip], color='green', zorder=5, marker='s')
    plt.annotate("Inflection Point", xy=(ip, y_ip), xytext=(ip - 3, y_ip - 12),
                 arrowprops=dict(arrowstyle='->', color='green'), color='green')

plt.title("Annotated Curve Sketch")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
````

This is the payoff of the whole month: a single graph that shows exactly where the function rises, falls, and bends — all derived from algebra, not guesswork.
