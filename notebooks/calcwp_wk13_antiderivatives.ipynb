## 0. Welcome to Integral Calculus

Over Weeks 1–12, we built up **Differential Calculus**: given a function, find its rate of change. Starting this week, we flip the question around: **given a rate of change, can we recover the original function?**

This reverse process is called **antidifferentiation**, or **integration** — and it's just as central to Calculus as differentiation was. In fact, the two are deeply connected (we'll see exactly how in Week 15, with the Fundamental Theorem of Calculus). For now, think of integration simply as "undoing" a derivative.


## 1. What Is an Antiderivative?

A function $F(x)$ is an **antiderivative** of $f(x)$ if:

$$ F'(x) = f(x) $$

**Example:** Is $F(x) = x^3$ an antiderivative of $f(x) = 3x^2$?

Check by differentiating $F$: $ \dfrac{d}{dx}[x^3] = 3x^2 = f(x) $ ✓ Yes!


````
from sympy import symbols, diff

x = symbols('x')
F = x**3
f = diff(F, x)
print("If F(x) =", F, ", then F'(x) =", f)
````

**Here's the twist:** $F(x) = x^3$ isn't the *only* antiderivative of $3x^2$. What about $F(x) = x^3 + 5$?

````
F2 = x**3 + 5
print("F'(x) for F(x) = x^3 + 5:", diff(F2, x))
````

Same derivative! Since the derivative of any **constant** is 0, adding any constant to an antiderivative doesn't change its derivative. This means every function has **infinitely many** antiderivatives, all differing by a constant. We write the general antiderivative using an arbitrary constant $C$:

$$ \int 3x^2 \, dx = x^3 + C $$

The symbol $\int \cdots dx$ is called an **indefinite integral**, and finding it is exactly the process of antidifferentiation.

## 2. The Power Rule for Integration

Since differentiation and integration are reverses of each other, there's a "reverse power rule":

$$ \int x^n \, dx = \frac{x^{n+1}}{n+1} + C \quad (\text{for } n \neq -1) $$

In words: **increase the exponent by 1, then divide by the new exponent.**

**Example:** $ \int x^5 \, dx = \dfrac{x^6}{6} + C $

Let's compute this with `sympy.integrate()`, which handles the "+C" implicitly (you add it yourself when writing the final answer, since `sympy` just gives *one* antiderivative).


````
from sympy import integrate

result = integrate(x**5, x)
print("∫x^5 dx =", result, "+ C")
````

## 3. Sum Rule and Constant Multiple Rule

Just like differentiation, integration distributes over sums/differences, and constants can be pulled out:

$$ \int [f(x) \pm g(x)] \, dx = \int f(x)\,dx \pm \int g(x)\,dx $$
$$ \int c \cdot f(x) \, dx = c \int f(x) \, dx $$

This means we can integrate a polynomial **term by term**, exactly the way we differentiated polynomials term by term back in Week 6.

**Example:** $ \int (4x^3 - 6x^2 + 2x - 5) \, dx $


````
expr = 4*x**3 - 6*x**2 + 2*x - 5
result = integrate(expr, x)
print("Integral:", result, "+ C")

# Verify by differentiating back
print("Check:", diff(result, x))
````

## 4. A Few More Building Blocks

**Integrating a constant:** $ \int c \, dx = cx + C $ (since the derivative of $cx$ is just $c$).

**A quick preview:** `sympy` also knows how to integrate trigonometric, exponential, and other functions you may encounter later — even though we haven't covered the rules formally, `integrate()` handles them automatically, the same way `diff()` did back in Week 6.


````
from sympy import sin, cos, exp

print("∫ 7 dx =", integrate(7, x), "+ C")
print("∫ cos(x) dx =", integrate(cos(x), x), "+ C")
print("∫ exp(x) dx =", integrate(exp(x), x), "+ C")
print("∫ 1/x dx =", integrate(1/x, x), "+ C")
````

Notice $ \int \frac{1}{x}\,dx = \ln(x) $ — this is the **one exception** to the power rule: when $n=-1$ (i.e., $x^{-1} = 1/x$), the "increase exponent, divide" rule would require dividing by zero, so integration gives a natural logarithm instead. You don't need to derive this yourself — just recognize that `sympy` (and later coursework) will handle it correctly.

## 5. Finding the Specific Constant $C$

An indefinite integral gives a whole **family** of antiderivatives (one for each value of $C$). If we're given one additional piece of information — an **initial condition** — we can solve for the exact value of $C$ that fits our specific situation.

**Example:** Suppose $ f'(x) = 6x^2 - 4 $, and we know $ f(1) = 5 $. Find $f(x)$.

**Step 1:** Integrate to get the general form:


````
from sympy import symbols, integrate, Symbol

C = symbols('C')
f_prime = 6*x**2 - 4
f_general = integrate(f_prime, x) + C
print("f(x) =", f_general)
````

**Step 2:** Use the initial condition $f(1)=5$ to solve for $C$.

````
from sympy import Eq, solve

equation = Eq(f_general.subs(x, 1), 5)
C_value = solve(equation, C)[0]
print("C =", C_value)

f_specific = f_general.subs(C, C_value)
print("f(x) =", f_specific)
````
