## 1. The Power Rule

For any function of the form $ f(x) = x^n $:

$$ f'(x) = n \cdot x^{n-1} $$

**Example:** If $ f(x) = x^3 $, then $ f'(x) = 3x^2 $.

Let's verify this with `sympy`, using `symbols` and `diff` (similar to how we used `limit` in Week 4).


````
from sympy import symbols, diff

x = symbols('x')
f = x**3

f_prime = diff(f, x)
print(f_prime)
````

`sympy` confirms: $ \frac{d}{dx}[x^3] = 3x^2 $, exactly matching the power rule.

The power rule also applies to constants and constant multiples:
- $ \frac{d}{dx}[c] = 0 $ (the derivative of a constant is always 0 — a flat line has zero slope)
- $ \frac{d}{dx}[c \cdot x^n] = c \cdot n \cdot x^{n-1} $


````
expr1 = 5              # a constant
expr2 = 4 * x**3         # a constant multiple

print("d/dx[5] =", diff(expr1, x))
print("d/dx[4x^3] =", diff(expr2, x))
````

## 2. The Sum and Difference Rule

Derivatives distribute across addition and subtraction:

$$ \frac{d}{dx}[f(x) \pm g(x)] = f'(x) \pm g'(x) $$

This means you can differentiate a polynomial **term by term**.

**Example:** $ f(x) = x^3 - 4x^2 + 5x - 7 $


````
expr = x**3 - 4*x**2 + 5*x - 7
print(diff(expr, x))
````

## 3. The Product Rule

When two functions are **multiplied** together, you *cannot* just differentiate each piece separately. Instead:

$$ \frac{d}{dx}[f(x) \cdot g(x)] = f'(x) \cdot g(x) + f(x) \cdot g'(x) $$

**Example:** $ f(x) = x^2 $ and $ g(x) = x + 3 $, so we want $ \frac{d}{dx}[(x^2)(x+3)] $.

By hand:
- $ f'(x) = 2x $, $ g'(x) = 1 $
- $ \frac{d}{dx}[f \cdot g] = 2x(x+3) + x^2(1) = 2x^2 + 6x + x^2 = 3x^2 + 6x $

Let's check with `sympy`:


````
expr = (x**2) * (x + 3)
print(diff(expr, x))
````

`sympy` returns the same result (possibly in a different but equivalent form) as our hand calculation — confirming the product rule.

## 4. The Quotient Rule

For a function that's one expression **divided** by another:

$$ \frac{d}{dx}\left[\frac{f(x)}{g(x)}\right] = \frac{f'(x)g(x) - f(x)g'(x)}{[g(x)]^2} $$

**Example:** $ \dfrac{x^2}{x+1} $


````
expr = x**2 / (x + 1)
print(diff(expr, x))
````

`sympy` gives the result directly. You can also ask `sympy` to `simplify()` the output to make it more readable:

````
from sympy import simplify

result = diff(expr, x)
print("Unsimplified:", result)
print("Simplified:  ", simplify(result))
````

## 5. The Chain Rule

Used when one function is **nested inside** another — a "function of a function."

$$ \frac{d}{dx}\big[f(g(x))\big] = f'(g(x)) \cdot g'(x) $$

**Example:** $ (x^2 + 1)^5 $ — here the "outer" function is (something)$^5$, and the "inner" function is $x^2+1$.

By hand: bring down the power, reduce it by one, then multiply by the derivative of the inside:
$$ \frac{d}{dx}\big[(x^2+1)^5\big] = 5(x^2+1)^4 \cdot (2x) = 10x(x^2+1)^4 $$


````
expr = (x**2 + 1)**5
result = diff(expr, x)
print(simplify(result))
````

**A note on more advanced functions:** `sympy` also knows the derivatives of trigonometric, exponential, and logarithmic functions (e.g. `sin(x)`, `exp(x)`, `log(x)`) — even though we haven't covered those rules explicitly, `diff()` handles them automatically if you ever need them.

````
from sympy import sin, exp

expr1 = sin(x**2)      # chain rule with sine
expr2 = exp(3*x)        # chain rule with exponential

print("d/dx[sin(x^2)] =", diff(expr1, x))
print("d/dx[exp(3x)]  =", diff(expr2, x))
````

## 6. Recognizing Which Rule Applies

A common source of confusion for beginners: knowing *which* rule to reach for.

**Power rule** — use when you see a single power: $ x^n $

**Sum/difference rule** — use when terms are added or subtracted: $ f(x) \pm g(x) $ (differentiate term by term)

**Product rule** — use when two expressions are multiplied: $ f(x) \cdot g(x) $

**Quotient rule** — use when one expression is divided by another: $ \dfrac{f(x)}{g(x)} $

**Chain rule** — use when something is "inside" another function, like a power applied to an expression: $ f(g(x)) $

Many real problems combine **more than one rule at once** — for example, a quotient where the numerator itself needs the product rule. `sympy.diff()` handles all of these automatically, which is exactly why it's such a useful tool once the underlying rules make sense to you.
